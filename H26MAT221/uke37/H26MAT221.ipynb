# MAT221 Diskret Matematikk Høst 2026

## **Velkommen!**

I MAT221 får du en innføring i tallteori, grafteori, kombinatoriske design og i teorien for opptelling. 

Teorien og modellene i dette kurset er nyttige verktøy som kan brukes til å forstå og beskrive mange fenomener av diskret natur innen naturvitenskap, samfunnsvitenskap, språkvitenskap og datastrukturer, samt andre fenomener av allmenn og hobbymatematisk interesse. Derfor gir emnet et spennende og nyttig grunnlag for videre studier i matematikk og informatikk.

Her finner du **emnebeskrivelsen**: [MAT221](https://www4.uib.no/studier/emner/mat221)
