```python
def pascal_enplusy(n):
    """
    Generate a Pascal triangle for LaTeX embedding.
    Sends output to the file-like object ``out`` (default: sys.stdout).
    """
    res = r'\begin{matrix}' + '\n'
    for i, row in enumerate(pascal(n)):
        res += f'(1 + y)^{i} & = &'
        res += '&' * (n-i)
        res += ' &+& '.join (f'{coeff:2d} y^{j}' if j else f'{coeff:2d}' for j, coeff in enumerate(row))
        res += (r'\\' + '\n')
    res += (r'\end{matrix}' + '\n')
    return res
```
