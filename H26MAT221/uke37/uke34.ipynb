## Multiplikasjonsprinsippet.

La oss si at en aktivitet består av $k$ hendelser $H_1, H_2, \dots, H_k$.

Dersom hendelsene $H_1, H_2, \dots, H_k$ har henholdsvis $\alpha_1, \alpha_2, \dots, \alpha_{k}$ uavhengige utfall i følgende forstand:

For alle de $\alpha_1$ utfallene som skjer i hendelse $H_1$ kan alle de $\alpha_2$ utfallene i hendelse $H_2$ skje og så videre.

Da kan aktiviteten utføres på $\alpha_1 \alpha_2 \cdots \alpha_k$ måter.

### Eksempel

Du vil plassere tre personer på tre forskjellige markerte plasser. 

Tenk på hendelse $H_1$ som at du plasserer den første personen, $H_2$ at du plasserer den andre personen og $H_3$ at du plasserer siste personen.

Da har $H_1$ tre utfall, og uavhengig av disse har $H_2$ to utfall og $H_1$ har ett utfall. 

Du kan altså utføre denne aktiviteten på seks forskjellige måter.

Samme tankemåte viser at du kan plasserer fire personer på fire forskjellige plasser på $4! = 4 \cdot 3 \cdot 2 \cdot 1$ måter.

### Eksempel

Du vil plassere to personer på fire forskjellige plasser. Tenker vi på plassering av første personen som et utfall $H_1$ og plassering av andre personen som et utfall $H_2$ har vi fire utfall for $H_1$ og tre utfall for $H_2$.

Denne aktiviteten kan altså utføres på $4 \cdot 3 = 12$ forskjellige måter.

## Teorem

La $\binom nr$ være antallet av måter vi kan velge en samling av $r$ forskjellige objekter fra en samling bestående av $n$ forskjellige objekter.

Da er 
$$\binom nr = \frac{n!}{r!(n-r)!}$$

### Begrunnelse:

Vi forklarte egentlig dette med eksemplet der vi valgte fire ut av tyve plasser.

#### Mere formell begrunnelse:
Hvis vi vet at teoremet holder for $n$ da kan vi forklare hvorfor det også må holde for $n+1$:

$$\binom{n+1}r = \binom nr + \binom n{r-1} = \frac{n!}{r!(n-r)!} + \frac{n!}{(r-1)!(n-r+1)!}$$

Som på side 6 i læreboken kan dette uttrykket algebraisk reduseres til $\frac{(n+1)!}{r!(n+1-r)!}$.

Dette kan illustreres ved å skrive linjene over hverandre:

De indre tallene i den nederste linjen er gitt ved summen av de to tallene til venstre og til høyre for dem i linjen over.

```python
# Kode som kan produsere Pascals trekant. Koden er ikke viktig å forstå!

def pascal(n):
    """
    Yield up to row ``n`` of Pascal's triangle, one row at a time.
    The first row is row 0.
    """
    if not isinstance (n, int):
        raise TypeError ('n must be an integer')
    if n < 0:
        raise ValueError ('n must be an integer >= 0')

    def newrow(row):
        "Calculate a row of Pascal's triangle given the previous one." 
        prev = 0
        for x in row:
            yield prev + x
            prev = x
        yield 1

    prevrow = [1]
    yield prevrow
    for x in range(n):
        prevrow = list(newrow(prevrow))
        yield prevrow

def pascal_latex(n):
    """
    Generate a Pascal triangle for LaTeX embedding.
    """
    res = r'\begin{matrix}' + '\n'
    for i, row in enumerate(pascal(n)):
        res += '&' * (n-i)
        res += ' && '.join ('%2d' % coeff for coeff in row)
        res += (r'\\' + '\n')
    res += (r'\end{matrix}' + '\n')
    return res
    
```

### Utdrag av Pascals trekant

```python
display(Latex(pascal_latex(7)))
```

```python
# Kode for ny figur. Ignorer koden...

def xyterm(i, j, coeff):
    if j == 0:
        if i > 0:
            return f'x^{i}'
        else:
            return '1'
    elif j == i:
        return f'y^{i}'
    else:
        return f'{coeff:2d} x^{i-j}y^{j}'
        
def pascal_xplusy(n):
    """
    Generate a Pascal triangle for powers of x + y
    """
    res = r'\begin{matrix}' + '\n'
    for i, row in enumerate(pascal(n)):
        res += f'(x + y)^{i} & = &'
        res += '&' * (n-i)
        res += ' &+& '.join (xyterm(i, j, coeff) for j, coeff in enumerate(row))
        res += (r'\\' + '\n')
    res += (r'\end{matrix}' + '\n')
    res = res.replace("(x + y)^1", "x + y")
    res = res.replace("^1", "")
    res = res.replace("(x + y)^0", "1")
    return res
```
