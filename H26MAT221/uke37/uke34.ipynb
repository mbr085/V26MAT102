## Multiplikasjonsprinsippet.

La oss si at en aktivitet består av $k$ hendelser $H_1, H_2, \dots, H_k$ som kan ha henholdsvis $\alpha_1, \alpha_2, \dots, \alpha_{k}$ uavhengige utfall i den forstand at hva som skjer i hendelse $H_1$ ikke har innflytelse på hva som skjer i hendelse $H_2$ og så videre.

Da kan aktiviteten utføres på $\alpha_1 \alpha_2 \cdots \alpha_k$ måter.

### Eksempel

Du vil plassere tre personer på tre forskjellige markerte plasser. 

Tenk på hendelse $H_1$ som at du plasserer den første personen, $H_2$ at du plasserer den andre personen og $H_3$ at du plasserer siste person.

Da har $H_1$ tre utfall, og uavhengig av disse har $H_2$ to utfall og $H_1$ har ett utfall. 

Du kan altså utføre denne aktiviteten på seks forskjellige måter.

Samme tankemåte viser at du kan plasserer fire personer på fire forskjellige plasser på $4! = 4 \cdot 3 \cdot 2 \cdot 1$ måter.

### Eksempel

Du vil plassere to personer på fire forskjellige plasser. Tenker vi på plassering av første personen som et utfall $H_1$ og plassering av andre personen som et utfall $H_2$ har vi fire utfall for $H_1$ og tre utfall for $H_2$.

Denne aktiviteten kan altså utføres på seks forskjellige måter.

## Teorem

Aktiviteten å plassere $n$ forskjellige objekter på $n$ markerte plasser kan utføres på $n!$ forskjellige måter.

## Korollar
Aktiviteten å plassere $n$ forskjellige objekter i rekkefølge kan utføres på $n!$ forskjellige måter.

## Kombinatorisk eksplosjon

Tallet $n!$ vokser veldig fort med $n$:

```python
# beregn n! for n mellom 1 og 14
resultat = [1]
for n in range(2, 15):
    resultat.append(n * resultat[-1])
```

```python
resultat
```

### Eksempel

Fire personer plasseres på hver sin plass. Det er 20 markerte plasser å velge mellom. 

La oss analysere hvor mange måter det kan gjøres på. Første person kan plasseres på 20 forskjellige plasser. Andre personer på 19 plasser, tredje person på 18 plasser og siste person på 17 plasser. Resultatet blir

$20 \cdot 19 \cdot 18 \cdot 17 = \frac{20!}{16!}$

```python
import math

math.factorial(20) // math.factorial(16) 
```

## Utvalg

Hvis vi i stedet for fire personer velger å markere fire av tyve plasser med identiske merkelapper kan vi tenke at vi først husker rekkefølgen av plasseringen av merkelappene og deretter glemmer den. Når vi husker rekkefølgen blir det like mange måter å utføre aktiviteten på som hvis vi hadde plassert personer. Når vi glemmer rekkefølgen er det $4!$ forskjellige rekkefølger som gir samme plasseringen av lappene fordi fire lapper kan plasseres i rekkefølge på $4!$ måter. 

Resultatet blir altså
$$ \frac{20!}{4! 16!}.$$

Dette tallet skriver vi $\binom{20}{4}$. Vi sier "tyve-velg-fire".

```python
math.comb(20, 4), math.factorial(20) // (math.factorial(4) * math.factorial(16))
```
