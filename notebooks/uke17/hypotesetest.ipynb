# To utvalgs t-test

# Oppgave

En butikkjede måler effekten av plassering av godteri i butikkene. 

Femti tilfeldig utvalgte butikker (z-butikkene) plasserer godteriet ved inngangen til butikken.

Femti tilfeldig utvalgte butikker (y-butikkene) beholdt godteriet i standard plasseringen rett før kassene.

Vi vil teste hypotesen om salget av godteri falt ved å flytte på plasseringen til godteriet.

Vi gjør nullhypotesen at salget ikke endret seg med den alternative hypotesen at salget av godteri falt ved å endre plassering:

**Antatte gjennomsnittsverdier** $\mu_y$ og $\mu_z$

**Nullhypotese ($H_0)$:** $\mu_y = \mu_z$.

**Alternativ hypotese ($H_a$):** $\mu_y > \mu_z$.

Salget av godteri i butikkene i prøveperioden i KNOK, det til si i tusinder av kroner, fordelte seg slik:

$t$-verdien for en to utvalgs test er:

$$t = \frac{E(y) - E(z)}{\sigma(y, z) \sqrt{2/n}}.$$

Her er $E(y)$ gjennomsnittet til $y$-verdiene og $\sigma(y, z)$ er det kombinerte standardavviket
$$\sigma(y,z) = \sqrt{\frac{\sigma(y)^2 + \sigma(z)^2}{2}}.$$

Husk gjerne at standardavviket $\sigma(y)$ kan beregnes ved formelen $\sigma(y)^2 = E(y^2) - E(y)^2$.

## Vil teste:
**Antatte gjennomsnittsverdier** $\mu_y$ og $\mu_z$

**Nullhypotese ($H_0)$:** $\mu_y = \mu_z$.

**Alternativ hypotese ($H_a$):** $\mu_y > \mu_z$.

```python
Ey = np.mean(y)
Ey2 = np.mean(y**2)
Ez = np.mean(z)
Ez2 = np.mean(z**2)
```

```python
sy = np.sqrt(Ey2 - Ey**2)
sz = np.sqrt(Ez2 - Ez**2)
```

```python
sy, sz
```

```python
syz = np.sqrt((Ey2 - Ey**2 + Ez2 - Ez**2)/2)
```

```python
syz
```

```python
t = (Ey - Ez)/(np.sqrt(2/len(y)) * syz)
```

## To utvalgs t-test

La $n$ være antallet av observasjoner i hver av utvalgene. For oss er $n = 50$.

La $E(y)$ være gjennomsnittet til tallene $y_i$ og la $E(z)$ være gjennomsnittet til tallene $z_i$. 

Vi kommer igjen til å gi en formel for en $t$-verdi som du for eksempel [finner på Wikipedia sin side om t-test i avsnittet "**Equal sample sizes and variance**".](https://en.wikipedia.org/wiki/Student's_t-test)

For å skrive formelen litt fint vil jeg innføre det kombinerte standardavviket
$$\sigma(y,z) = \sqrt{\frac{\sigma(y)^2 + \sigma(z)^2}{2}}.$$

Husk gjerne at standardavviket $\sigma(y)$ er gitt ved formelen $\sigma(y)^2 = E(y^2) - E(y)^2$.

$t$-verdien for vår to utvalgs test er:

$$t = \frac{E(y) - E(z)}{\sigma(y, z) \sqrt{2/n}}.$$

Antallet av frihetsgrader er $2n - 2$.
